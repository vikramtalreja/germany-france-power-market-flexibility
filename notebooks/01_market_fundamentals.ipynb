```python
from pathlib import Path
import sys
ROOT = Path.cwd() if (Path.cwd() / 'src').exists() else Path.cwd().parent
sys.path.insert(0, str(ROOT / 'src'))
from market_flex.prices import compare
prices, summary = compare(ROOT / 'data/raw')
summary
```

```python
import matplotlib.pyplot as plt
monthly = prices.tz_convert('Europe/Berlin').resample('MS').mean()
monthly[['DE-LU','FR']].plot(figsize=(10,4), ylabel='EUR/MWh', title='Monthly mean prices, 2024')
plt.show()
```

DE-LU averages €78.51/MWh; France €58.02/MWh. Negative-price hours: 457 and 352. Mean spread: €20.49/MWh. Prices alone do not establish causes.
